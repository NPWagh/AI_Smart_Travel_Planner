```python
from google import genai
from getpass import getpass

api_key = getpass("Enter your Gemini API Key (press Enter to skip): ")

client = None

if api_key.strip():
    client = genai.Client(api_key=api_key)
    print("Gemini AI configured successfully!")
else:
    print("Gemini API skipped. Basic travel planning mode will be used.")
```

```python
print("====================================")
print("       AI SMART TRAVEL PLANNER")
print("====================================")

destination = input("Enter your travel destination: ")

days = int(input("Enter number of travel days: "))

budget = float(input("Enter your total budget in INR: "))

travellers = int(input("Enter number of travellers: "))

interests = input(
    "Enter your interests (e.g. Nature, Adventure, Food, Shopping, History): "
)

travel_type = input(
    "Enter travel type (Family, Friends, Solo, Couple): "
)

print("\nTravel details collected successfully!")
```

```python
def generate_travel_plan(destination, days, budget, travellers,
                         interests, travel_type):

    prompt = f"""
    Act as an intelligent AI travel planning assistant.

    Create a personalized travel plan using these details:

    Destination: {destination}
    Duration: {days} days
    Total Budget: ₹{budget}
    Number of Travellers: {travellers}
    Interests: {interests}
    Travel Type: {travel_type}

    Include the following sections:

    1. Travel Overview
    2. Day-wise Travel Itinerary
    3. Popular Tourist Attractions
    4. Food and Restaurant Suggestions
    5. Accommodation Suggestions
    6. Estimated Budget Breakdown:
       - Accommodation
       - Food
       - Local Transportation
       - Sightseeing and Activities
       - Other Expenses
    7. Important Travel Tips
    8. Best Time to Visit

    Requirements:
    - Keep the plan within the given budget as far as possible.
    - Consider the number of travellers.
    - Customize suggestions according to the user's interests.
    - Provide a practical day-wise schedule.
    - Clearly mention that prices and opening hours should be verified.
    - Do not invent confirmed hotel bookings or ticket availability.
    - Use simple and clear English.
    """

    if client:
        try:
            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=prompt
            )
            return response.text

        except Exception as e:
            print("Gemini API error:", e)
            print("Switching to basic travel planning mode.")

    return basic_travel_plan(destination, days, budget, travellers, interests)


def basic_travel_plan(destination, days, budget, travellers, interests):

    daily_budget = budget / days

    plan = f"""
    AI SMART TRAVEL PLAN
    ====================

    Destination: {destination}
    Duration: {days} Days
    Travellers: {travellers}
    Total Budget: ₹{budget:,.2f}
    Daily Budget: ₹{daily_budget:,.2f}
    Interests: {interests}

    DAY-WISE ITINERARY
    -----------------
    """

    for day in range(1, days + 1):
        plan += f"""
        Day {day}:
        - Explore popular attractions in {destination}.
        - Enjoy local food and cultural experiences.
        - Spend time exploring places related to: {interests}.
        - Keep some time for relaxation.
        """

    plan += f"""

    BUDGET BREAKDOWN (SUGGESTED)
    ---------------------------
    Accommodation: 30% = ₹{budget * 0.30:,.2f}
    Food: 20% = ₹{budget * 0.20:,.2f}
    Transportation: 20% = ₹{budget * 0.20:,.2f}
    Sightseeing: 20% = ₹{budget * 0.20:,.2f}
    Other Expenses: 10% = ₹{budget * 0.10:,.2f}

    TRAVEL TIPS
    -----------
    1. Check hotel and transport prices before booking.
    2. Carry valid identity documents.
    3. Check local weather conditions.
    4. Keep emergency contact details.
    5. Verify attraction timings before visiting.

    Note: This is a basic sample plan. It does not contain
    verified local attractions, hotel availability, or live prices.
    """

    return plan
```

```python
travel_plan = generate_travel_plan(
    destination,
    days,
    budget,
    travellers,
    interests,
    travel_type
)

print("\n====================================")
print("       YOUR AI TRAVEL PLAN")
print("====================================\n")

print(travel_plan)
```

```python
from google.colab import files

file_name = "AI_Smart_Travel_Plan.txt"

with open(file_name, "w", encoding="utf-8") as file:
    file.write(travel_plan)

print("Travel plan saved successfully!")

files.download(file_name)
```
